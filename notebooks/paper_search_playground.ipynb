# Paper Search Playground

Manual notebook for testing paper retrieval by query using `ResearchService.search_academic_papers()`.

```python
from pathlib import Path
import os
import sys

repo_root = Path.cwd()
if not (repo_root / "src").exists():
    repo_root = repo_root.parent

if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.services.research_service import ResearchService
```

```python
query = "what changed in attention mechanism from attention is all you need to now"
max_papers = 8

# Optional debug flags
os.environ["DEBUG_QUERY_EXPANSION"] = "true"
os.environ["DEBUG_CITATIONS"] = "true"
```

```python
service = ResearchService(storage_dir=str(repo_root / "data"))
papers = await service.search_academic_papers(query=query, max_papers=max_papers)
len(papers)
```

```python
rows = []
for idx, p in enumerate(papers, start=1):
    rows.append({
        "rank": idx,
        "source": p.source,
        "title": p.title,
        "citations": p.citations,
        "published_date": p.published_date,
        "url": p.url,
        "pdf_url": p.pdf_url,
    })

try:
    import pandas as pd
    display(pd.DataFrame(rows))
except Exception:
    for row in rows:
        print(f"{row['rank']:>2}. [{row['source']}] {row['title']}")
        print(f"    citations={row['citations']} published={row['published_date']}")
        print(f"    url={row['url']}")
        print(f"    pdf={row['pdf_url']}")
        print()
```
